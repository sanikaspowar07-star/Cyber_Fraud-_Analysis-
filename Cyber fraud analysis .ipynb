```python
total_loss = df["Financial_Loss"].sum()
print("Total Financial Loss:", round(total_loss, 2))
```

#Find the total loss by cybercrime type 

```python
crime_loss = df.groupby("Crime_Type")["Financial_Loss"].sum()
crime_loss = crime_loss.sort_values(ascending=False)
print(crime_loss)
```

#Find the State with the highest financial Loss 

```python
state_loss = df.groupby("State")["Financial_Loss"].sum()
state_loss = state_loss.sort_values(ascending=False)
print(state_loss)
```

STEP 4:Visualization using Python 

```python
sns.set_theme(style="whitegrid")
```

#Bar Chart:Complaints by crime Type 

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="Crime_Type",
    order=df["Crime_Type"].value_counts().index)
plt.title("Number of Complaints by Cybercrime Type")
plt.xlabel("Cybercrime Type")
plt.ylabel("Number of Complaints")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

#Bar Chart Financial Loss By Crime Type 


```python
crime_loss = (
    df.groupby("Crime_Type")["Financial_Loss"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

crime_loss.plot(kind="bar")

plt.title("Total Financial Loss by Cybercrime Type")
plt.xlabel("Cybercrime Type")
plt.ylabel("Total Financial Loss")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

#Pie Chart:Complaint Status Distribution 

```python
status_counts = df["Complaint_Status"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Complaint Status Distribution")
plt.show()
```

#Histogram:Victim Age Distribution 

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Victim_Age"].dropna(),
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Cyber Fraud Victim Ages")
plt.xlabel("Victim Age")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

#Line Chart:Monthly Financial Loss 

```python
df["Month"] = df["Complaint_Date"].dt.to_period("M")

monthly_loss = (
    df.groupby("Month")["Financial_Loss"]
    .sum()
    .sort_index()
)

print(monthly_loss)
```

```python
plt.figure(figsize=(12, 5))

monthly_loss.plot(kind="line", marker="o")

plt.title("Monthly Financial Loss from Cyber Fraud")
plt.xlabel("Month")
plt.ylabel("Total Financial Loss")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
